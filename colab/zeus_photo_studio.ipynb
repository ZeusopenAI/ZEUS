## 🎨 CÔNG CỤ 1: IOPaint (Lama Cleaner) — Xóa Icon Búng Tim & Chi Tiết Thừa
- **Thao tác trên điện thoại:** Cực mượt, nút to, dùng ngón tay tô màu cọ lên icon búng tim rồi ấn nút chạy là ảnh phẳng phiu như chưa từng có icon.
- **Chỉ cần bấm nút chạy cell bên dưới:** Chờ 1-2 phút, click vào link `https://xxxx.trycloudflare.com` được in ra.

```python
# [1] KHỞI ĐỘNG IOPAINT (LAMA CLEANER)
import subprocess, threading, time, re

print("⏳ Đang cài đặt thư viện IOPaint...")
subprocess.run("pip install -q iopaint", shell=True)

print("🚀 Đang khởi động máy chủ IOPaint (Model LaMa SOTA)...\n")
iopaint_proc = subprocess.Popen("iopaint start --model=lama --device=cuda --port=8080 --host=0.0.0.0", shell=True)
tunnel_proc = subprocess.Popen("cloudflared tunnel --url http://127.0.0.1:8080", shell=True, stderr=subprocess.PIPE, text=True)

def watch_iopaint_url():
    for line in tunnel_proc.stderr:
        match = re.search(r'https://[a-zA-Z0-9-]+\.trycloudflare\.com', line)
        if match:
            url = match.group(0)
            print("="*65)
            print(f"🎉 LINK TRUY CẬP IOPAINT CHO ĐIỆN THOẠI:\n👉 {url} 👈")
            print("="*65)
            print("💡 Hướng dẫn: Chạm vào link trên, tải ảnh lên, lấy ngón tay quẹt cọ lên icon búng tim rồi bấm nút chạy.")
            break

threading.Thread(target=watch_iopaint_url, daemon=True).start()
```
